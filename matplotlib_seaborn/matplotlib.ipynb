```python
import pandas as pd
sales_data = pd.read_csv('sales_data_large.csv')
sales_data
```

```python
# plot total sales by product

total_sales_by_product = sales_data.groupby("Product Category")["Units Sold"].sum()


total_sales_by_product
```

```python
total_sales_by_product.plot(kind='bar', color='blue', )
plt.show()
```

```python
# plot a sales trend by date
sales_trend = sales_data.groupby("Date")["Total Revenue"].sum().reset_index()


sales_trend
```

```python
plt.plot(sales_trend["Date"], sales_trend["Total Revenue"])
plt.show()
```
