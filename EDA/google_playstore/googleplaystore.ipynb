```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

# Suppress all warning messages during execution to keep output clean
warnings.filterwarnings("ignore")

# Display matplotlib plots directly below cells instead of in a separate window
%matplotlib inline
```

```python
df=pd.read_csv('googleplaystore.csv')
df.head()
df.drop(10472,inplace=True)
```

```python
df.at[10472, "Reviews"] = "3000000"
df[~df["Reviews"].str.isnumeric()]
```

```python
df['Reviews']=df["Reviews"].astype(int)
```

```python
df.info()
```

```python
df[~df["Size"].str.contains("M") & ~df["Size"].str.contains("k") ]["Size"].unique()
# df[df["Size"] == "Varies with device"]
# df[df["Size"] == "1,000+"]
```

```python
def convert_size(size):
    size = str(size).strip().lower()
    if size.endswith("m"):
        return float(size[:-1])

    if size.endswith("k"):
        return float(size[:-1])/1024

    if size == "Varies with device":
        return np.nan
```

```python
df["Size"] = df["Size"].apply(convert_size)
df.head()
```

```python
df['Size'].describe()
```

```python
df['Installs'].unique()     
```

```python
df.dropna(subset=['Installs'], inplace=True)
```

```python
df.isnull().sum()   
```

```python
df['Installs'].unique()
```

```python
df['Price'].unique()
```

```python
chars_to_remove=["+",",","$"]
cols_to_clean=["Price","Installs"]

for item in chars_to_remove:
    for cols in cols_to_clean:
        df[cols] = df[cols].str.replace(item,"")

df.head()
```

```python
df['Installs']=df['Installs'].astype(int)
df['Price']=df['Price'].astype(float)
df.info()
```

```python
df['Last Updated'] = pd.to_datetime(df['Last Updated'])
df['Last Updated'].head()
```

```python
df.head()
```

```python

df["day"] = df['Last Updated'].dt.day
df["month"] = df['Last Updated'].dt.month
df["year"] = df['Last Updated'].dt.year
df.head()
```

```python
df.to_csv('google_cleaned.csv')
```

```python
df = pd.read_csv("google_cleaned.csv")
df.info()
```

```python
df[df.duplicated("App")]
```

```python
df = df.drop_duplicates(subset=["App"],keep="first")
```

```python
df[df.duplicated("App")]
```

```python
df.info()
```

```python
numeric_features = [feature for feature in df.columns if df[feature].dtype != 'str'] # O for object
categorical_features = [feature for feature in df.columns if df[feature].dtype == 'str']

# print columns
print('We have {} numerical features : {}'.format(len(numeric_features), numeric_features))
print('\nWe have {} categorical features : {}'.format(len(categorical_features), categorical_features))
```

```python
## Proportion of count data on categorical columns
for col in categorical_features:
    print(df[col].value_counts(normalize=True)*100)
    print('---------------------------')
```

```python
plt.figure(figsize=(15,15))
plt.title("Distribution of numericcal Features", fontsize=20)


for i in range(0, len(numeric_features)):
    plt.subplot(5, 3, i+1)
    sns.kdeplot(data=df, x=numeric_features[i], fill=True)
    plt.xlabel(numeric_features[i], fontsize=12)
    plt.tight_layout()
```

```python
# categorical columns
plt.figure(figsize=(20, 15))
plt.suptitle('Univariate Analysis of Categorical Features', fontsize=20, fontweight='bold', alpha=0.8, y=1.)
category = [ 'Type', 'Content Rating']
for i in range(0, len(category)):
    plt.subplot(2, 2, i+1)
    sns.countplot(x=df[category[i]],palette="Set2")
    plt.xlabel(category[i])
    plt.xticks(rotation=45)
    plt.tight_layout() 
```
