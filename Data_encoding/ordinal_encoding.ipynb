```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import OrdinalEncoder

size_df = pd.DataFrame({"shirt_size": ["S", "M", "L", "XL", "XXL"]})
encoder = OrdinalEncoder(categories=[["S", "M", "L", "XL", "XXL"]],dtype=np.int64)
size_encoded = pd.DataFrame()
size_encoded = encoder.fit_transform(size_df[["shirt_size"]])
size_encoded
```
