```python
import pandas as pd
import numpy as np
```

```python
df= pd.DataFrame({
    "job_title": ["Data Scientist", "Data Analyst", "Data Engineer", "Data Scientist", "Data Analyst"],
    "salary": [120000, 80000, 100000, 130000, 85000]     })

df
```

```python
mean_salary = df.groupby("job_title")["salary"].mean().sort_values().to_dict()
```

```python
df["salary_label"] = df["job_title"].map(mean_salary)
df[["job_title", "salary", "salary_label"]]
```
